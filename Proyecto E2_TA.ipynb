### RANKING ELO

```python
url = "https://tennisabstract.com"
```

```python
driver = webdriver.Chrome()
driver.get(url)
```

```python
try:
    rank_elo = driver.find_element(By.LINK_TEXT, "... All Women's Elo Ratings")
    enlace = rank_elo.get_attribute("href")
    driver.close()

except Exception as e:
    print("Error al obtener los torneos:", e)
```

```python
print(enlace)
```

```python
driver = webdriver.Chrome()
driver.get(enlace)
```

```python
filas = driver.find_elements(By.CSS_SELECTOR, "#reportable tbody tr")

for fila in filas:
    columnas = fila.find_elements(By.TAG_NAME, "td")

    if len(columnas) > 1:
        ranking = columnas[0].text
        nombre = columnas[1].text
        elo = columnas[3].text
        dura = columnas[6].text
        arcilla = columnas[8].text
        hierba = columnas[10].text

        print(f'{ranking}) {nombre}: {elo} , dura: {dura}, arcilla: {arcilla}, hierba: {hierba}')
```
